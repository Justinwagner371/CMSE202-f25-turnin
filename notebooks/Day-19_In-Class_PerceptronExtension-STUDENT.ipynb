```python
# Put code here to start and generate new cells as needed.
import pandas as pd
df = pd.read_csv('sonar.csv')
df.head()
```

Copy your percentron class to the cell below. **Note** sklearn has a **Perceptron** function. We should avoid using the same function name of your perceptron class.

```python
# put your code here

import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

df["Class"] = df["Class"].map({"Rock": -1, "Mine": 1})
labeled_data = df.values.tolist()

class Percentron():
    def __init__(self, labeled_data):        
        self.data = labeled_data
        feature_length = len(labeled_data[0]) - 1  
        self.weights = np.zeros(feature_length + 1) 

    def predict(self, feature_set):
        x = np.append(feature_set, 1) 
        result = np.dot(self.weights, x)
        return 1 if result > 0 else -1

    def train(self, max_epochs=100):
        for epoch in range(max_epochs):
            errors = 0
            for row in self.data:
                features, label = row[:-1], row[-1]
                prediction = self.predict(features)
                if prediction != label:
                    x = np.append(features, 1)
                    self.weights += label * x
                    errors += 1
            if errors == 0:
                break 
```

Train your percentron class with sonar data. What's the accuracy?

```python
# put your code here
train_data, test_data = train_test_split(labeled_data, test_size=0.2, random_state=42)

model = Percentron(train_data)
model.train()

y_true = [row[-1] for row in test_data]
y_pred = [model.predict(row[:-1]) for row in test_data]
accuracy = accuracy_score(y_true, y_pred)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```
