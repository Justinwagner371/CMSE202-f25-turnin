```python
import matplotlib.pyplot as plt
import numpy as np

L = np.linspace(0, 100, 100)

supply = 0.5 * L + 10
demand = -0.5 * L + 60
h = 10
demand_post = demand - h

plt.figure(figsize=(8, 6))
plt.plot(L, supply, label='Supply (Type A)')
plt.plot(L, demand, label='Demand (Pre-Mandate)')
plt.plot(L, demand_post, label='Demand (Post-Mandate)', linestyle='--')
plt.title('Labor Market for Type A Workers')
plt.xlabel('Labor (L)')
plt.ylabel('Wage')
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 6))
plt.plot(L, supply, label='Supply (Type B)')
plt.plot(L, demand, label='Demand (Pre-Mandate)')
plt.plot(L, demand_post, label='Demand (Post-Mandate)', linestyle='--')
plt.title('Labor Market for Type B Workers')
plt.xlabel('Labor (L)')
plt.ylabel('Wage')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

L = np.linspace(0, 100, 100)
supply = 0.5 * L + 10
demand = -0.5 * L + 60
h = 10
demand_post = demand - h
supply_post = supply - h/2

plt.figure(figsize=(8, 6))
plt.plot(L, supply, label='Supply (Pre-Mandate)')
plt.plot(L, demand, label='Demand (Pre-Mandate)')
plt.plot(L, supply_post, label='Supply (Post-Mandate)', linestyle='--')
plt.plot(L, demand_post, label='Demand (Post-Mandate)', linestyle='--')
plt.title('Labor Market: Firms Cannot Distinguish Worker Types')
plt.xlabel('Labor (L)')
plt.ylabel('Wage')
plt.legend()
plt.grid(True)
plt.show()
```
